# Dataset Exploration
Use this notebook to inspect the Kaggle Resume Dataset after downloading it into `data/raw/`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("../data/raw/data/data")

print("Dataset path:", DATA_DIR)
print("Dataset exists:", DATA_DIR.exists())
```

```python
category_counts = {}

for category_dir in DATA_DIR.iterdir():
    if category_dir.is_dir():
        category_counts[category_dir.name] = len(
            list(category_dir.glob("*.pdf"))
        )

category_df = pd.DataFrame(
    list(category_counts.items()),
    columns=["Category", "Resume_Count"]
).sort_values("Resume_Count", ascending=False)

category_df
```
